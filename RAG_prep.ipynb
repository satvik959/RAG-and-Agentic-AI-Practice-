```python
import os
from getpass import getpass
from llama_index.core import SimpleDirectoryReader, VectorStoreIndex, Settings
from llama_index.core.node_parser import SentenceSplitter
from llama_index.embeddings.openai import OpenAIEmbedding
from llama_index.llms.openai import OpenAI


class RAG:
    def __init__(self, data):
        self.data = data
        Settings.embed_model = OpenAIEmbedding(model='text-embedding-3-small')  #taking the embedding model
        Settings.llm = OpenAI(model='gpt-4o-mini', temperature=0)  #usage of llm and temperature marks the randomness,high temp indicates creativity

    def load_documents(self):
        self.documents = SimpleDirectoryReader(self.data).load_data()

    def split_documents(self):
        splitter = SentenceSplitter(chunk_size=1024, chunk_overlap=20)
        self.nodes = splitter.get_nodes_from_documents(self.documents)

    def create_index(self):
        self.index = VectorStoreIndex(self.nodes)
        self.query_engine = self.index.as_query_engine(similarity_top_k=2)

    def ask(self, question):
        return self.query_engine.query(question)


if __name__ == "__main__":
    if not os.getenv("OPENAI_API_KEY"):
        os.environ["OPENAI_API_KEY"] = getpass("OpenAI API key: ")

    os.makedirs("data", exist_ok=True)

    if not os.path.exists("data/notes.txt"):
        with open("data/notes.txt", "w", encoding="utf-8") as file:
            file.write("""
RAG stands for Retrieval-Augmented Generation.
RAG retrieves relevant information before generating an answer.
Chunking divides large documents into smaller pieces.
Chunk overlap helps preserve context across adjacent chunks.
Embeddings represent text as numerical vectors.
""")

    rag = RAG("data")
    rag.load_documents()
    rag.split_documents()
    rag.create_index()

    answer = rag.ask("Why is chunk overlap useful?")
    print(answer)
```

```python
import math
import numpy as np
import torch

from scipy.spatial.distance import cdist
from sentence_transformers import SentenceTransformer


model = SentenceTransformer( "sentence-transformers/paraphrase-MiniLM-L6-v2",
    device="cpu")
embedding = model.encode(documents)
print("Shape",embedding.shape)
print("the top 10 embeddings are :")
print(embedding[0][:10])#top 10 vectors for similarity
```

```python
def euclidian_distance(vector1,vector2):
  square_sum = sum((float(x)-float(y))**2
              for x , y in zip(vector1,vector2))
  return math.sqrt(square_sum)
print(euclidian_distance(embedding[0],embedding[1]))
```
